# **Linear Regression Maths**

---

### **1) Model (Line Equation):**

For one feature:
$$
h(x) = wx + b
$$
For multiple features:
$$
h(x) = w_1 x_1 + w_2 x_2 + \cdots + w_n x_n + b
$$
Compact (vector) form:
$$
h(x) = w^T x + b
$$

Where,
* h(x): prediction
* w: weights
* b: bias
